#### Vraag 3c: De oppervlaktelading van een geleidende schijf op constant potentiaal

De oppervlakte lading $\sigma(r)$ van een geleidende schijf met straal $R$ is niet uniform verdeeld over de schijf, maar wordt gegeven door (Problems 2.58 and 2.63 Griffiths 5th edition):
$$\sigma(r) = \frac{Q}{2 \pi R \sqrt{R^2 - r^2}}$$

1. Bereken de totale lading op de schijf met behulp van vraag 3a en 3b.
2. Vergelijk de oppervlakte lading in het centrum van de geleidende schijf met dat van een uniform geladen schijf met dezelfde totale lading.
3. Plot de oppervlakte lading van de geleidende schijf met de berekende lading $Q$, tussen $r = 0$ en $R$.
4. Plot in hetzelfde figuur de oppervlaktelading van een uniform geladen schijf met dezelfde lading $Q$.


````python
###
###Jouw code hier###
R = 3  # Straal van de schijf in meters
Q = 1e-14  # Totale lading (Coulomb)

# Straalwaarden
r_values = np.linspace(0, R, 1000)

# Oppervlaktelading voor de geleidende schijf (formule uit de vraag)
sigma_values = Q / (2 * np.pi * R * np.sqrt(R**2 - r_values**2))

# Oppervlaktelading voor een uniforme schijf (constante waarde)
sigma_uniform = Q / (np.pi * R**2)

# Plot beide oppervlakteladingen
plt.figure()
plt.plot(r_values, sigma_values, label=r'$\sigma(r)$ voor de geleidende schijf')
plt.axhline(sigma_uniform, color='r', linestyle='--', label=r'$\sigma_{\text{uniform}}$ voor de uniforme schijf')
plt.xlabel('Straal r (m)')
plt.ylabel('Oppervlaktelading $\sigma(r)$ (C/m²)')
plt.title('Vergelijking van oppervlaktelading voor geleidende en uniforme schijf')
plt.legend()
plt.grid(True)
plt.show()
###
````

4. Schrijf een python functie die de relevante parameters als input neemt en de capaciteit van een cilinder condensator als output heeft.

````python

def vlakkeplaat(epsilon_r,epsilon0,surface, distance):
    C = (epsilon_r*surface*epsilon0)/distance
    return C

def cilinder(epsilon_r, epsilon0, length, straal_binnen, straal_buiten):
    C = (2*np.pi*epsilon_r*epsilon0*length)/np.log((straal_buiten/straal_binnen))
    return  C

````
