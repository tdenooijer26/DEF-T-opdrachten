## *Opdracht 3*: Foto van de schets van jouw meetopstelling in.

Het schema stelt een spanningsdeler voor die bestaat uit een krachtgevoelige weerstand R(F) en een vaste weerstand R₂. De krachtgevoelige weerstand R(F) is aangesloten tussen de voedingsspanning van 5 V en het meetpunt. De vaste weerstand R₂ is verbonden tussen het meetpunt en massa (GND). Het analoge meetsignaal wordt afgenomen op het knooppunt tussen R(F) en R₂ en is aangesloten op de analoge ingang A0 van de Arduino.

De uitgangsspanning $V_{out}$ op A0 wordt bepaald door de spanningsdelerformule:

$$
V_{out} = V_{in} \cdot \frac{R_2}{R(F) + R_2}
$$

Aangezien R₂ een vaste weerstand is en R(F) varieert afhankelijk van de uitgeoefende kracht, zal de uitgangsspanning veranderen wanneer de kracht op de sensor toeneemt of afneemt. Bij toenemende kracht neemt de weerstand R(F) af. Hierdoor verandert de spanningsverdeling en dus ook de uitgangsspanning volgens bovenstaande relatie.

## *Opdracht 6*: Meten en fitten, Leerdoel

```python
# Zet hier jouw data in een voor jou handig formaat
import numpy as np
import matplotlib.pyplot as plt
arduinodata = np.array([497,626,686,765,832,822,846])
massadata = np.array([56,100,153,226,312,384,480])
```

```python
# Plot hier de meting en fit een lijn door de metingen.
# Maak gebruik van wat je geleerd hebt bij het practicum.
%matplotlib inline

from scipy.optimize import curve_fit

P0=(0,0,0)

def modelfunctie(x,a,b,c):
    return a*np.exp(b*x)+c

Popt,pcov = curve_fit(modelfunctie,arduinodata,massadata,p0=P0)
a,b,c = Popt

bits_array = np.arange(np.min(arduinodata),np.max(arduinodata))

plt.figure()
plt.plot(arduinodata,massadata,".",label="data")
plt.plot(bits_array,modelfunctie(bits_array,*Popt),"--", label="curve_fit")
plt.xlabel("bits")
plt.ylabel("massa [g]")
plt.legend()
plt.show()

#print("m= %2.f e^(%2.f )")
```

## *Opdracht 7*: Iteratie nodig?


In de beginopstelling functioneerde de sensor, maar de gemeten bits bleven bij lichte voorwerpen bij 0. Dit betekent dat de uitgangsspanning nauwelijks veranderde en dat de Arduino geen onderscheid kon maken tussen kleine massas. Bij het plaatsen van zwaardere voorwerpen werden wel verschillende meetwaarden geregistreerd, wat aangeeft dat het systeem alleen gevoelig was bij grotere krachten op de sensor.

De spanningsdelerformule verteld ons dat $V_{out}$ gevoeliger wordt voor de verandering van $R(F)$ bij het verlagen van $R_2$. De oorspronkelijke weerstand was te groot voor het meetbereik van kleine krachten, waardoor het systeem pas reageerde bij hogere belastingen. De iteratie die is toegepast is het plaatsen van meer weerstanden parallel aan $R_2$. Hierdoor wordt de totale waarde van $R_2$ verlaagd.

Gezien ook met deze opstelling de arduino niet gevoelig was voor kleine massaverschillen, is de opstelling niet verder verandert, maar is bij een tweede iteratie gekozen voor het gebruiken van grotere massa's. Hier erkent de arduino wel verschillende massa's en kunnen de meetwaarden worden gebruikt voor een curve-fit.
