## *Opdracht 3*: Foto van de schets van jouw meetopstelling in.

Het schema stelt een spanningsdeler voor die bestaat uit een krachtgevoelige weerstand R(F) en een vaste weerstand R₂. De krachtgevoelige weerstand R(F) is aangesloten tussen de voedingsspanning van 5 V en het meetpunt. De vaste weerstand R₂ is verbonden tussen het meetpunt en massa (GND). Het analoge meetsignaal wordt afgenomen op het knooppunt tussen R(F) en R₂ en is aangesloten op de analoge ingang A0 van de Arduino.

De uitgangsspanning $V_{out}$ op A0 wordt bepaald door de spanningsdelerformule:

$$
V_{out} = V_{in} \cdot \frac{R_2}{R(F) + R_2}
$$

Aangezien R₂ een vaste weerstand is en R(F) varieert afhankelijk van de uitgeoefende kracht, zal de uitgangsspanning veranderen wanneer de kracht op de sensor toeneemt of afneemt. Bij toenemende kracht neemt de weerstand R(F) af. Hierdoor verandert de spanningsverdeling en dus ook de uitgangsspanning volgens bovenstaande relatie.
